## Abstract

Mental health challenges among young adults, particularly anxiety, depression, and insomnia have become a pressing public health concern. Mental health resources are inaccessible to some due to cost and stigma leading youth to turn to music as a coping mechanism. Our project investigates which music genres exhibit the strongest positive and negative associations with self-reported mental health. Along with genres, whether tempo (BPM) is a meaningful predictor of psychological distress. 

We analyzed the Music & Mental Health Survey Results dataset from Kaggle (n = 736), which contains reported anxiety, depression, and insomnia scores (0-10 scale), along with favorite music genre, daily listening hours, and preferred BPM. Via IQR filtering, we sifted out missing values and outliers. What was left was a usable sample of ~700 observations. We conducted both univariate and bivariate analyses, where we characterized distributions and compared genre groups against mental health outcomes.

Noteworthy patterns were found among the data. Lo-fi listeners had some of the highest median scores across all three mental health categories (anxiety, depression, and insomnia). Metal listeners showed higher insomnia and depression scores but lower anxiety. Most genres however, showed a wider interquartile range. This limits the predictability of mental health based on genre alone. Continuous variables such as daily listening hours and BPM had small correlations to mental health outcomes as well. This suggests that what someone listens to may matter more than how much or how fast. Correlation does not imply causation, therefore our findings should be interpreted with caution.

## Research Question

Which music genre exhibits the strongest positive and negative associations with a person’s mental health profile, specifically across self-reported Anxiety, Depression, and Insomnia scores on a 0-10 scale? 

This project makes use of statistical inference to determine if identifying a specific genre as one’s favorite and its average bpm correlates with significantly higher or lower mental health scores. By controlling age and hours of daily listening, we will isolate the relationship between musical preference and mental health to ensure that we are strictly analyzing the correlation between the two. This analysis aims to uncover whether certain genre preferences can be used as indicators of psychological distress patterns.  

## Background and Prior Work

Mental health among college students has become an increasingly urgent public health concern, anxiety, depression, and chronic stress rates being reported in recent years. In a 2024 report by the American College Health Association, 60% of college students reported overwhelming anxiety while another portion reported being functionally limited due to depression [1]. The research topic is necessary to see how college students utilize music as a low-barrier mental health service to cope with the daily stressors of life. Many college students do not seek formal mental health resources due to monetary constraints and the stigma around these struggles, and music has emerged as a unique outlet for emotional regulation. Listening to music has transformed from passive listening to an active tool for stress management and crisis aversion. Furthermore, identifying the relationship between music genres and mental health can provide insights into how different musical content can either contribute to or detract from mental health struggles when used as a tool for self-remedy. In recent years, studies have shown that music is beneficial to students with high stress levels. For example, a 2025 study titled “Statistical Insights into the Influence of Music on Mental Health” highlights the impact of musical engagement on psychological well-being by giving quantitative evidence [2].


Unlike therapy or medication, music requires no appointment, no cost, and no disclosure. Music is a low-barrier outlet for students, as it is already embedded in their daily life. The use of music as an emotional regulator serves as a prompt for researchers to look beyond music as entertainment but rather a legitimate coping mechanism. A thesis from the University of Guelph by Parya Abadeh argues that while music is a regular tool for emotional processing, the type of engagement determines whether the outcome is beneficial or not. Abadeh explains that “certain songs-especially those with negative or ambiguous emotional content- can reinforce rumination and other maladaptive emotion-regulation patterns, ultimately elevating stress…” [3]. While this study affirms that music listening broadly correlates with positive mental health outcomes, it treats musical engagement as a relatively uniform behavior, leaving open the question of whether what students listen to matters as much as the fact that they are listening at all.


However, the correlation between them is debated profusely, especially regarding certain genres of music and their levels of emotional intensity and calmness. A 2022 study by Kuo Wang, Sunyu Gao, and Jianhao Huang titled “Learning About Your Mental Health From Your Playlist? Investigating the Correlation Between Music Preference and Mental Health of College Students” found that preference for the pop music genre indicated less reported mental health struggles while heavy music preference showed an inverse relationship. Their results indicated the “correlation was significant and positive between pop music and mental health.” [4] They also observed a “significant and inverse relationship between college students’ preference for heavy music and their mental health”. [4] These findings are compelling, but leave some questions unanswered, since it remains unclear whether genre preference has a causative or consequential relationship to psychological stress patterns. Additionally, it is also unclear whether listening frequency of certain genres can play a role in regulating or destabilizing mental health struggles. With these posed questions, this project aims to address these gaps by holistically examining the relationship between student listening habits and mental health struggles to see how music can be utilized as a tool or deterrent for these emotional hardships.

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Loading Data
raw_path = 'data/00-raw/mxmh_survey_results.csv'
df = pd.read_csv(raw_path)

# Verifying tidyness: 
# This dataset is tidy because each row is a single respondent and each column is a single variable.
# We are also checking for potential duplicate entries.
# We verify this by checking for duplicate entries.
print(f"Number of duplicate rows: {df.duplicated().sum()}")
display(df.head())

# Data selection
# We are keeping these 7 columns as they are the only variables needed to satisfy our hypothesis. 
relevant_cols = ['Age', 'Hours per day', 'BPM', 'Fav genre', 'Anxiety', 'Depression', 'Insomnia']
df = df[relevant_cols]
```

```python
# Missing Data Handling

print(f"Initial Dataset Shape: {df.shape}")
initial_count = len(df)
bpm_missing = df['BPM'].isnull().sum()

# Calculate pre-cleaning missing data
missing_data_pre = df.isnull().sum()

# Drop rows missing critical labels to create df_clean
df_clean = df.dropna(subset=['BPM', 'Anxiety', 'Depression', 'Insomnia', 'Fav genre']).copy()

# Calculate post-cleaning missing data
missing_data_post = df_clean.isnull().sum()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left Plot: Pre-cleaning
missing_data_pre.plot(kind='bar', color='salmon', edgecolor='black', ax=axes[0])
axes[0].set_title('Missing Values (Pre-Cleaning)')
axes[0].set_ylabel('Number of Missing Entries')
axes[0].tick_params(axis='x', rotation=45)

# Right Plot: Post-cleaning
missing_data_post.plot(kind='bar', color='lightgreen', edgecolor='black', ax=axes[1])
axes[1].set_title('Missing Values (Post-Cleaning)')
axes[1].set_ylabel('Number of Missing Entries')
axes[1].set_ylim(0, 5) 
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Outlier Removal using IQR 

def remove_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filter the dataframe to only include rows within the bounds
    return dataframe[(dataframe[column] >= lower_bound) & (dataframe[column] <= upper_bound)]

# Check distributions before IQR filtering
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.boxplot(x=df_clean['BPM'], color='skyblue')
plt.title('BPM Distribution (Pre-IQR Filter)')

plt.subplot(1, 2, 2)
sns.boxplot(x=df_clean['Age'], color='salmon')
plt.title('Age Distribution (Pre-IQR Filter)')
plt.show()

# Apply the IQR filter to BPM and Age
df_clean = remove_outliers_iqr(df_clean, 'BPM')
df_clean = remove_outliers_iqr(df_clean, 'Age')

# Check distributions after IQR filtering
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.boxplot(x=df_clean['BPM'], color='skyblue')
plt.title('BPM Distribution (Post-IQR Filter)')

plt.subplot(1, 2, 2)
sns.boxplot(x=df_clean['Age'], color='salmon')
plt.title('Age Distribution (Post-IQR Filter)')

plt.tight_layout()
plt.show()
```

```python
# Summary and Results
print(f"Original rows: {initial_count}")
print(f"Rows with missing BPM: {bpm_missing}")
print(f"Final usable rows after all cleaning: {len(df_clean)}")

print("\n--- Summary Statistics for Key Variables ---")
display(df_clean.describe())
```

#### Interpretation of Univariate Distributions
The distributions of the ages are heavily right skewed. Considering this skew, we inferred that the demographic of our sample is teens and young adults. According to the distribution of listening time, the right skew indicates engagement ranges from 2 to 4 hours per day. The preferred BPM is closer to a normal distribution, which is an indication that the preferred tempos are strongly clustered at the center. The center clusters show a preferred BPM between 100-130 BPM. 
 
The distributions for our dependent variables are revealing and far from uniformly distributed. Anxiety scores are heavily skewed toward the higher end at scores of 7-10 with few responding below a score of 5, entailing that this specific sample has higher anxiety levels as less of an outlier and more of a baseline. The depression variable is somewhat bimodal with large peaks at both the lowest score, 0, and higher scores of 6-8, which suggests that the sample is divided between those who report no depressive symptoms at all and those who are on the more extreme end of the spectrum. Furthermore, relatively few individuals report moderate depression scores between these values. Insomnia is right-skewed with most respondents clustering towards the lower end for insomnia scores, showing that extreme issues with sleeping problems are less universal than anxiety or depression. However, the distribution remains notably spread across scores between 2-6, indicating that moderate levels of sleep difficulty is still fairly common in this sample.

Understanding these baselines is critical before comparing across groups. Given that high anxiety is the norm in this sample while depression is, conversely, sharply polarized, elevated scores in later analyses should not be automatically attributed to genre-specific effects. Instead, they may be attributed more to the generalized mental health struggles of the sample. Insomnia is right-skewed with the majority of respondents clustered at score 0, showing that this sample does not severely struggle with sleep. However, the even distribution spread across moderate scores suggests that mild-moderate sleep issues is consistent and requires careful contextualization. Additionally, the strong clustering of preferred BPM between 100-130 also means that tempo-based comparisons may have limited variation since more extreme BPM preferences are underrepresented.

## Discussion and Conclusion

Mental health among young adults has become an increasingly urgent public health concern, and music listening remains as a universal tools for emotional regulation. This project examines the relationship between specific music genre preferences and their correlations with reported mental health outcomes for Anxiety, Depression, and Insomnia. These mental health struggles were measured on a 0-10 severity scale where higher scores reflect more severe symptoms. Our central research question asks what genres exhibit the strongest positive and negative associations with someone’s mental wellbeing. To explore this, we conducted univariate analyses of demographic and mental health variables, bivariate comparisons of genre preferences against distress scores, and correlation analyses examining roles of continuous variables like BPM and daily listening hours in predicting mental health outcomes. 

Our analysis revealed multiple noteworthy patterns that correlations do exist between genre preferences and mental health outcomes. Lofi listeners reported the highest median scores across all three mental health dimensions while Classical and Gospel genre listeners reported the lowest median distress scores overall. Interestingly, Metal listeners showed a counterintuitive pattern, as they had elevated Depression and Insomnia scores with a below-average Anxiety median, suggesting that high energy music genres may function as an outlet to cope with Anxiety instead of triggering and worsening this mental health struggle. However, the wide IQR’s observed across nearly all genres indicate substantial individual variability, entailing that genre preference alone is nota  strong predictor of an individual's mental health status. Additionally, BPM and hours of daily listening showed little to no meaningful correlation with mental health scores, challenging our original beliefs that tempo would be a relevant attributor of high-arousal symptoms like Insomnia and Anxiety. This aligns with Wang etc. al.’s (2002) finding that genre preference rather than the listening behavior itself is a more meaningful predictor of mental health outcomes.

There are important limitations to highlight and acknowledge. To begin with, the dataset is relatively small (n = 736) and was distributed primarily through social media platforms like Reddit and Discord, skewing the sample toward younger, tech fluent populations. This population likely overrepresents music enthusiasts. Additionally, all mental health scores are self-reported and not clinically validated, introducing an element of subjectivity and possible response bias. Since the data was collected our data through self-reported surveys, no causal claims can be made since individuals with existing mental health struggles might gravitate towards certain genres instead of the music itself being the impetus of distress. This work suggests that music listening habits could have meaningful, correlative relationships with psychological well-being. Future work could expand on these findings and benefit from large, demographically diverse samples and utilize longitudinal study designs that track mental health changes over time. Additionally, clinically validated mental health measures instead of self-reported metrics can reduce subjectivity and increase the validity of findings. By utilizing clinical standards for mental health measures, future research can ensure that new resources being created are grounded in medically informed knowledge. Understanding these relationships can inform accessible, low-cost mental health resources for individuals who face resource barriers to formal care. These findings can particularly benefit college students who face disproportionate rates of anxiety and depression, as this demographic often avoids receiving formal care due to budget constraints and stigma around mental health struggles. If genre preference can serve as a meaningful indicator of psychological distress, music listening data can one day inform the development of mental health tools to create resources that have low-entry points to reach people who aren’t likely to seek formal support.
